## 6. Create MySQL Table

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS Users (
    id INT,
    name VARCHAR(100),
    email VARCHAR(150),
    postalZip VARCHAR(50),
    phone VARCHAR(50)
)
"""

cursor.execute(create_table_query)
print("Table created")
```

## 7. Load JSON Data into MySQL

```python
insert_query = """
INSERT INTO Users
(id, name, email, postalZip, phone)
VALUES (%s, %s, %s, %s, %s)
"""

cursor.execute("TRUNCATE TABLE Users")
cursor.executemany(insert_query, json_data["data"])
connection.commit()

print("Data loaded into MySQL")
```

## 8. Read Data from MySQL

```python
cursor.execute("SELECT * FROM Users")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

## 9. Load MySQL Data into Pandas DataFrame

The data is read from MySQL and stored in a Pandas DataFrame.

```python
cursor.execute("SELECT * FROM Users")
rows = cursor.fetchall()
columns = [column[0] for column in cursor.description]

df = pd.DataFrame(rows, columns=columns)
df
```

## 10. Display All Data

```python
display(df)
```

## 11. Change Email Addresses

Change the domain of every email address to `gmail.com`.

```python
df["email"] = df["email"].str.replace(r"@.*", "@gmail.com", regex=True)
df[["email"]]
```

## 12. Convert postalZip to Integer

Remove letters and other non-numeric characters, then convert the values to integers.

```python
df["postalZip"] = (
    df["postalZip"]
    .astype(str)
    .str.replace(r"\D", "", regex=True)
    .astype(int)
)

df[["postalZip"]]
```

## 13. Phone Number Conversion Function

Remove formatting, take two digits at a time, and convert them to ASCII. Values below 65 are replaced with `O`.

```python
def phone_to_ascii(phone):
    digits = ''.join(char for char in str(phone) if char.isdigit())
    result = ''

    for i in range(0, len(digits) - 1, 2):
        number = int(digits[i:i + 2])

        if number < 65:
            result += 'O'
        else:
            result += chr(number)

    return result
```

## 14. Test the Phone Function

```python
print(phone_to_ascii("(816) 530-4269"))
print(phone_to_ascii("1-811-920-9732"))
```

## 15. Apply Phone Conversion to the DataFrame

```python
df["phone"] = df["phone"].apply(phone_to_ascii)
df[["phone"]]
```
