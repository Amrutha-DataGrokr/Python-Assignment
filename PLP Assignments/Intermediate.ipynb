
## Q2) Topic: OOP: Classes, `__init__`, Inheritance & Polymorphism 
### Build a class hierarchy for a library system: 
(a) Base class LibraryItem: title, author, year, describe() method. 

````python
class LibraryItem:
    def __init__(self,title,author,year):
        self.title=title
        self.author=author
        self.year=year

    def describe(self):
        return f"{self.title} by {self.author} ({self.year})"
````

(b) Subclass Book: add pages and genre, override describe().

````python
class Book(LibraryItem):
    def __init__(self,title,author,year,pages,genre):
        super().__init__(title,author,year)
        self.pages=pages
        self.genre=genre

    def describe(self):
        return f"{self.title} by {self.author} ({self.year}) - {self.pages} pages, {self.genre}"
````

(c) Subclass Magazine: add issue_number and is_monthly, override describe().

````python
class Magazine(LibraryItem):
    def __init__(self,title,author,year,issue_number,is_monthly):
        super().__init__(title,author,year)
        self.issue_number=issue_number
        self.is_monthly=is_monthly

    def describe(self):
        monthly="Monthly" if self.is_monthly else "Not Monthly"
        return f"{self.title} by {self.author} ({self.year}) - Issue {self.issue_number}, {monthly}"
````

(d) Create a list with one Book and one Magazine, loop and call describe() -- demonstrate polymorphism.

````python
items=[
    Book("Python Basics","John Smith",2024,350,"Programming"),
    Magazine("Tech World","Priya Rao",2025,12,True)
]

for item in items:
    print(item.describe())
````


## Q3) Topic: Decorators & Context Managers 
### Implement the following two Python patterns: 
(a) Decorator @log_call: prints 'Calling <name>...' before and 'Done.' after. Apply to sum_to_n(n). 

````python
from functools import wraps

def log_call(func):
    @wraps(func)
    def wrapper(*args,**kwargs):
        print(f"Calling {func.__name__}...")
        result=func(*args,**kwargs)
        print("Done.")
        return result
    return wrapper

@log_call
def sum_to_n(n):
    return sum(range(1,n+1))

print(sum_to_n(5))
````

(b) Decorator @retry(times=3): re-calls function up to 3 times on Exception. Simulate random failure.

````python
import random
from functools import wraps

def retry(times=3):
    def decorator(func):
        @wraps(func)
        def wrapper(*args,**kwargs):
            for attempt in range(times):
                try:
                    return func(*args,**kwargs)
                except Exception as e:
                    print(f"Attempt {attempt+1} failed")
                    if attempt==times-1:
                        raise
        return wrapper
    return decorator

random.seed(1)

@retry(times=3)
def unstable():
    if random.random()<0.9:
        raise Exception("Random failure")
    return "Success"

try:
    print(unstable())
except Exception:
    print("Function failed after 3 attempts")
````

(c) Custom context manager TempFile: creates file on `__enter__`, deletes on `__exit__`. Write and read 3 lines.

````python
import os

class TempFile:
    def __init__(self,filename):
        self.filename=filename

    def __enter__(self):
        self.file=open(self.filename,"w+")
        return self.file

    def __exit__(self,exc_type,exc_value,traceback):
        self.file.close()
        os.remove(self.filename)

with TempFile("temp.txt") as file:
    file.write("Line 1\nLine 2\nLine 3\n")
    file.seek(0)
    print(file.read(),end="")
````


## Q4) Topic: pandas: read_csv, groupby, merge + numpy basics 
### You have students.csv (student_id, name, class) and scores.csv (student_id, subject, score): 
(a) Read both into DataFrames and merge on student_id. 

````python
import pandas as pd
import numpy as np

students=pd.DataFrame({
    "student_id":[1,2,3,4],
    "name":["Anna","Bob","Carly","Doug"],
    "class":["10A","10A","10B","10B"]
})
students.to_csv("students.csv",index=False)

scores=pd.DataFrame({
    "student_id":[1,1,2,2,3,3,4,4],
    "subject":["Math","Science","Math","Science","Math","Science","Math","Science"],
    "score":[80,90,70,85,60,75,95,88]
})
scores.to_csv("scores.csv",index=False)

students=pd.read_csv("students.csv")
scores=pd.read_csv("scores.csv")
merged=pd.merge(students,scores,on="student_id")
print(merged)
````

(b) Use groupby to find average score per subject and per student.

````python
avg_subject=merged.groupby("subject")["score"].mean()
avg_student=merged.groupby("name")["score"].mean()

print(avg_subject)
print(avg_student)
````

(c) Add a grade column: 'A' >= 85, 'B' >= 70, 'C' >= 55, 'F' otherwise.

````python
def get_grade(score):
    if score>=85:
        return "A"
    elif score>=70:
        return "B"
    elif score>=55:
        return "C"
    else:
        return "F"

merged["grade"]=merged["score"].apply(get_grade)
print(merged[["name","subject","score","grade"]])
````

(d) Use numpy: mean, standard deviation, 75th percentile of all scores.

````python
all_scores=merged["score"].to_numpy()

print("Mean:",np.mean(all_scores))
print("Standard Deviation:",np.std(all_scores))
print("75th Percentile:",np.percentile(all_scores,75))
````


## Q6) Topic: Exception Patterns & Clean Code 
### Refactor: def f(x,y): try: return x/y except: pass 
(a) Rename parameters, add a proper docstring (Args + Returns + Raises). 

````python
def safe_divide(dividend: float, divisor: float) -> float:
    """
    Safely divides two numbers.

    Args:
        dividend: The number to be divided.
        divisor: The number to divide by.

    Returns:
        The division result or None if division fails.

    Raises:
        ZeroDivisionError: If divisor is zero.
        TypeError: If inputs are not numbers.
    """
    try:
        return dividend / divisor

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
        return None

    except TypeError:
        print("Error: Inputs must be numbers.")
        return None
````

(c) Add type hints: (dividend: float, divisor: float) -> Optional[float].

````python
from typing import Optional

def safe_divide(dividend: float, divisor: float) -> Optional[float]:
    try:
        return dividend / divisor

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
        return None

    except TypeError:
        print("Error: Inputs must be numbers.")
        return None
````

(d) Write batch_divide(pairs: list[tuple]) calling safe_divide, skipping failures, logging warnings.

````python
import logging

logging.basicConfig(level=logging.WARNING)

def batch_divide(pairs: list[tuple]):
    results = []

    for dividend, divisor in pairs:
        result = safe_divide(dividend, divisor)

        if result is not None:
            results.append(result)
        else:
            logging.warning(f"Skipping {dividend} / {divisor}")

    return results


pairs = [(10, 2), (6, 0), (9, 3)]

print(batch_divide(pairs))
````
