## Q5) Topic: Comprehensions & Functional Patterns

Use list, dictionary, and set comprehensions along with `functools.reduce()` and `sorted()` with lambda.

Given a list of at least 6 employee dicts (name, dept, salary) across 3 departments. Use ONLY functional tools: 


````python
employees = [
    {"name": "Amrutha", "dept": "Engineering", "salary": 75000},
    {"name": "Rahul", "dept": "Engineering", "salary": 65000},
    {"name": "Priya", "dept": "HR", "salary": 55000},
    {"name": "Arjun", "dept": "HR", "salary": 62000},
    {"name": "Sneha", "dept": "Finance", "salary": 70000},
    {"name": "Kiran", "dept": "Finance", "salary": 58000}
]

print(employees)
````

(a) List comprehension: names of all 'Engineering' employees. 

````python
engg_employees = [
    employee["name"]
    for employee in employees
    if employee["dept"] == "Engineering"
]
print(engg_employees)
````

(b) Dict comprehension: {name: salary} for employees earning > 60000.

````python
high_earners = {
    employee["name"]: employee["salary"]
    for employee in employees
    if employee["salary"] > 60000
}

print(high_earners)
````

(c) Set comprehension: unique departments.

````python
departments = {
    employee["dept"]
    for employee in employees
}

print(departments)
````

(d) functools.reduce() to calculate total payroll. sorted() with lambda to print top 3 earners. 

````python
from functools import reduce

total_payroll = reduce(
    lambda total, employee: total + employee["salary"],
    employees,
    0
)

print("Total payroll:", total_payroll)
````

````python
top_3 = sorted(
    employees,
    key=lambda employee: employee["salary"],
    reverse=True
)[:3]

print("Top 3 earners:")

for employee in top_3:
    print(employee["name"], employee["salary"])
````
