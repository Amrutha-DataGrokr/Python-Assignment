# OOP Inheritance + Abstraction

This program demonstrates inheritance using `Person`, `Male`, and `Female` classes.

It also uses an Abstract Base Class (`ABC`) to prevent direct creation of `Person` objects.

```python
from abc import ABC, abstractmethod


class Person(ABC):

    @abstractmethod
    def get_gender(self):
        pass


class Male(Person):

    def get_gender(self):
        print("Male")


class Female(Person):

    def get_gender(self):
        print("Female")
```

## Create Objects of Child Classes

Both `Male` and `Female` can be instantiated because they implement the abstract `get_gender()` method.

```python
male = Male()
male.get_gender()

female = Female()
female.get_gender()
```

## Try to Instantiate `Person`

`Person` cannot be instantiated because it contains the abstract method `get_gender()`.

Python raises a `TypeError`.

```python
person=Person()  
```
